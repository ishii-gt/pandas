```python
import pandas as pd
```

```python
# 1. Load the dataset into a Pandas DataFrame.
df = pd.read_csv("airbnb_dirty_dataset.csv")
df
```

```python
# 2. Display the first 10 and last 10 rows.
print(df.head(10))
print(df.tail(10))
```

```python
# 3. Find the number of rows and columns.
print(df.sample)
```

```python
# 4. Display the data types of all columns.
print(df.dtypes)

```

```python
# 5. Count the missing (NaN) values in each column.
df.isnull().sum()
```

```python
# 6. Fill missing values in 'price' with the mean price.
df["price"] = df["price"].fillna(df["price"].mean())
```

```python
# 7. Fill missing values in 'rating' with the median rating.
# df["rating"].median()
df["rating"] = df["rating"].fillna(df["rating"].median())
```

```python
# 8. Fill missing values in 'city', 'property_type', and 'host' with the mode.
df["city"].mode()
df["city"]= df["city"].fillna(df["city"].mode()[0])
df["property_type"] .mode()
df["property_type"]= df["property_type"].fillna(df["property_type"].mode()[0])
df["host"] .mode()
df["host"] = df["host"].fillna(df["host"].mode()[0])
```

```python
# 9. Count duplicate rows
print(df.duplicated().sum())
```

```python
# 10. Remove all duplicate rows.
df = df.drop_duplicates()
```

```python
# 11. Show all listings where price > 50000.
print(df[df["price"] > 5000])
```

```python
# 12. Show listings where rating >= 4.5.
print(df[df["rating"] >= 4.5])
```

```python
# 13. Find the average price for each city.
print(df.groupby("city")["price"].mean())
```

```python
# 14. Find the top 5 most expensive listings.
top_5 = df.sort_values(by='price', ascending=False).head(5)

top_5
```

```python
# 15. Find listings with reviews > 200.
print(df[df["reviews"] > 200])
```

```python
# 16. Filter Apartment listings in Mumbai.
 
mumbai_apartments = df[
    (df["city"] == "Mumbai") &
    (df["property_type"] == "Apartment")
]

print(mumbai_apartments)
```

```python
# 17. Sort by price in descending order.
sorted_price = df.sort_values(by='price', ascending=False)

print(sorted_price)
```

```python
# 18. Create a new column 'price_per_night' (same as price).
df["price_per_night"] = df["price"]

df.head()
```

```python
# 19. Detect outliers in 'price' using the IQR method.

Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[(df["price"] < lower_limit) | (df["price"] > upper_limit)]

print(outliers)
```

```python
# 20. Remove the detected outliers from the dataset.
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Remove outliers
df = df[(df["price"] >= lower_limit) & (df["price"] <= upper_limit)]

print(df)
```

```python
# 21. Verify that no duplicates remain.
print(df.duplicated().sum())
```

```python
# 22. Verify that missing values have been handled.
print(df.isnull().sum())
```
